```python
import plotly.express as px
import pandas as pd

def create_side_by_side_plots(dataframe, column):
    # Bar plot
    bar_counts = dataframe[column].value_counts()
    bar_fig = px.bar(dataframe, x=bar_counts.index, y=bar_counts.values, title=f'Distribution of Countries Across {column}')
    
    # Pie chart
    pie_counts = dataframe[column].value_counts()
    pie_fig = px.pie(dataframe, names=pie_counts.index, values=pie_counts.values, title=f'Pie Chart of {column}', color_discrete_sequence=px.colors.qualitative.Plotly)
    
    # Display plots side by side
    bar_fig.show()
    pie_fig.show()

# Example usage
data = {
    'Hemisphere': ['Northern', 'Southern', 'Northern', 'Southern', 'Northern', 'Southern'],
    'Human Development Groups': ['High', 'Medium', 'Low', 'High', 'Low', 'Medium']
}

df = pd.DataFrame(data)
create_side_by_side_plots(df, 'Hemisphere')
create_side_by_side_plots(df, 'Human Development Groups')
```

```python
import pandas as pd
import numpy as np
from scipy.stats import linregress

# Load the DataFrame from the CSV file
df = pd.read_csv(r"C:\Users\Asus\OneDrive\Tài liệu\Desktop\data_sets\Life Expectancy at Birth.csv")

# Extract the years and life expectancy values
years = df.columns[7:].str.extract(r'(\d{4})')[0].astype(int)  # Extract the year from the column labels
life_expectancy_values = df.iloc[:, 7:]  # Select life expectancy columns

# Calculate the linear regression trend for life expectancy
trends = []
for idx, year in enumerate(years):
    slope, intercept, r_value, p_value, std_err = linregress(range(len(df)), life_expectancy_values.iloc[:, idx])
    trends.append(slope)

# Print the calculated trends for each year
for year, trend in zip(years, trends):
    print(f"Year: {year}, Trend: {trend}")

```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from statsmodels.tsa.arima.model import ARIMA
from sklearn.metrics import mean_squared_error

# Load the DataFrame from the CSV file
df = pd.read_csv(r"C:\Users\Asus\OneDrive\Tài liệu\Desktop\data_sets\Life Expectancy at Birth.csv")

# Calculate the mean life expectancy over all countries for each year
years = df.columns[7:].tolist()  # Exclude the non-year columns
mean_life_expectancy = df.iloc[:, 7:].mean()

# Create a datetime index
start_year = 1990
end_year = 2021
date_range = pd.date_range(start=f"{start_year}-01-01", end=f"{end_year}-12-31", freq='Y')

# Assign the datetime index to the mean_life_expectancy series
mean_life_expectancy.index = date_range

# Split the data into training and testing sets
train_size = int(len(mean_life_expectancy) * 0.8)
train_data, test_data = mean_life_expectancy[:train_size], mean_life_expectancy[train_size:]

# Fit an ARIMA model to the training data
order = (5, 1, 0)  # (p, d, q) values for ARIMA
model = ARIMA(train_data, order=order)
model_fit = model.fit()

# Forecast future life expectancy values
forecast_steps = len(test_data)
forecast_values = model_fit.forecast(steps=forecast_steps)

# Calculate the mean squared error of the forecast
mse = mean_squared_error(test_data, forecast_values)
print(f"Mean Squared Error: {mse:.2f}")

# Plot the actual vs. forecasted values
plt.figure(figsize=(10, 6))
plt.plot(test_data.index, test_data, label='Actual')
plt.plot(test_data.index, forecast_values, label='Forecast')
plt.xlabel('Year')
plt.ylabel('Life Expectancy')
plt.title('ARIMA Forecasting: Actual vs. Forecasted Life Expectancy')
plt.legend()
plt.show()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from statsmodels.tsa.holtwinters import ExponentialSmoothing

# Load the DataFrame from the CSV file
df = pd.read_csv(r"C:\Users\Asus\OneDrive\Tài liệu\Desktop\data_sets\Life Expectancy at Birth.csv")

# Calculate the mean life expectancy over all countries for each year
years = df.columns[7:].tolist()  # Exclude the non-year columns
mean_life_expectancy = df.iloc[:, 7:].mean()

# Create a datetime index
start_year = 1990
end_year = 2021
date_range = pd.date_range(start=f"{start_year}-01-01", end=f"{end_year}-12-31", freq='Y')

# Assign the datetime index to the mean_life_expectancy series
mean_life_expectancy.index = date_range

# Apply a moving average to the data
window_size = 5  # Adjust this parameter as needed
moving_avg = mean_life_expectancy.rolling(window=window_size).mean()

# Apply exponential smoothing to the data
exponential_model = ExponentialSmoothing(mean_life_expectancy, trend='add', seasonal='add', seasonal_periods=5)
exponential_fit = exponential_model.fit()

# Plot the original, moving average, and smoothed data
plt.figure(figsize=(10, 6))
plt.plot(mean_life_expectancy.index, mean_life_expectancy, label='Original Data')
plt.plot(mean_life_expectancy.index, moving_avg, label=f'{window_size}-Year Moving Average')
plt.plot(mean_life_expectancy.index, exponential_fit.fittedvalues, label='Exponential Smoothing')
plt.xlabel('Year')
plt.ylabel('Life Expectancy')
plt.title('Smoothing Techniques: Original vs. Moving Average vs. Exponential Smoothing')
plt.legend()
plt.show()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import zscore

# Load the DataFrame from the CSV file
df = pd.read_csv(r"C:\Users\Asus\OneDrive\Tài liệu\Desktop\data_sets\Life Expectancy at Birth.csv")

# Calculate the mean and standard deviation of life expectancy values
mean_life_expectancy = df.iloc[:, 7:].mean(axis=1)
std_life_expectancy = df.iloc[:, 7:].std(axis=1)

# Calculate the Z-Score for each country's life expectancy values
import numpy as np

z_scores = (
    df.iloc[:, 7:] - mean_life_expectancy.to_numpy()[:, np.newaxis]
) / std_life_expectancy.to_numpy()[:, np.newaxis]


# Define a threshold for anomaly detection (e.g., z_score > 2 or z_score < -2)
anomaly_threshold = 2

# Identify anomalies using the threshold
anomalies = np.abs(z_scores) > anomaly_threshold

# Plot anomalies over time
plt.figure(figsize=(10, 6))
plt.imshow(anomalies, cmap='coolwarm', aspect='auto', extent=[1990, 2021, 0, len(anomalies)])
plt.xlabel('Year')
plt.ylabel('Country')
plt.title('Anomaly Detection: Life Expectancy Anomalies Over Time')
plt.colorbar(label='Anomaly')
plt.show()
```
